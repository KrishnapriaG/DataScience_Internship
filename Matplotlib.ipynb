```python
Basic Plot
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
plt.plot(x, y)
plt.show()
```

```python
Plot Customizations
```

```python
plt.plot(x,y,color='green',linestyle='--',marker='o')
plt.title('My Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
Multiple Lines
```

```python
x=[1,2,3]
y1=[1,2,3]
y2=[1,2,3]
plt.plot(x,y1,label='Squares')
plt.plot(x,y2,label='linear')
plt.legend()
plt.show()
```

```python
Bar Chart
```

```python
categories=['A','B','C']
values=[5,7,3]
plt.bar(categories,values)
plt.show()
```

```python
Histogram
```

```python
data=[1,1,2,3,3,4,4,5]
plt.hist(data,bins=5)
plt.show()
```

```python
Scatter Plot
```

```python
x=[1,2,3,4]
y=[10,20,25,30]
plt.scatter(x,y)
plt.show()
```

```python
Subplots
```

```python
plt.subplot(1,2,1)
plt.plot([1,2,3],[1,2,3])
plt.subplot(1,2,2)
plt.plot([1,2,3],[3,2,1])
plt.show()
```
